```python
def gestionar_inventario_zapateria():
    """
    Función para registrar la venta de zapatos y controlar el inventario disponible.
    """
    print("--- 👟 SISTEMA DE CONTROL DE INVENTARIO DE ZAPATOS 👟 ---\n")

    # 1. Solicitar el inventario inicial asegurando que sea un número entero positivo
    while True:
        try:
            inventario = int(input("Cantidad inicial de pares de zapatos en tienda: "))
            if inventario > 0:
                break  # Salimos del bucle si el número es válido
            print("⚠️ El inventario inicial debe ser mayor a 0.")
        except ValueError:
            print("❌ Entrada inválida. Por favor, ingresa solo números enteros.")

    total_vendidos = 0
    total_transacciones = 0

    # 2. Bucle principal: se ejecuta mientras haya zapatos disponibles en inventario
    while inventario > 0:
        print(f"\n📦 Inventario actual disponible: {inventario} pares")

        try:
            vendidos = int(input("¿Cuántos pares se vendieron en esta transacción? "))

            # Caso A: Intenta vender una cantidad menor o igual a cero
            if vendidos <= 0:
                print("⚠️ La cantidad vendida debe ser mayor a 0.")

            # Caso B: Intenta vender más de lo que hay en existencia
            elif vendidos > inventario:
                print(f"❌ No hay suficiente stock. Intentas vender {vendidos}, pero solo quedan {inventario}.")

            # Caso C: Venta exitosa
            else:
                inventario -= vendidos  # Restamos las ventas al inventario
                total_vendidos += vendidos  # Acumulamos el total de zapatos vendidos
                total_transacciones += 1  # Sumamos una transacción realizada
                print(f"✅ ¡Venta registrada! Quedan {inventario} pares en bodega.")

        except ValueError:
            print("❌ Entrada no válida. Ingresa un número entero (ejemplo: 2).")

    # 3. Resumen final al agotar todo el stock
    print("\n" + "="*45)
    print("🎉 ¡INVENTARIO AGOTADO! Resumen de la jornada:")
    print(f"• Total de pares vendidos: {total_vendidos}")
    print(f"• Total de transacciones exitosas: {total_transacciones}")
    print("="*45)

# Ejecutamos la función para iniciar el programa
gestionar_inventario_zapateria()
```
