```python
inventario = int(input("Cantidad inicial de pares de zapatos: "))

total_vendidos = 0

transacciones = 0

while inventario > 0:

    vendidos = int(input("¿Cuántos pares se vendieron? "))

    if 0 < vendidos <= inventario:

        inventario -= vendidos

        total_vendidos += vendidos

        transacciones += 1

        print(f"Inventario restante: {inventario}")

    else:

        print("Cantidad no válida. Inténtalo nuevamente.")

print(f"Total de pares vendidos: {total_vendidos}")

print(f"Total de transacciones: {transacciones}")
```
